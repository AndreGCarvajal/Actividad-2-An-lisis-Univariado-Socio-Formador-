```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el archivo de Excel
archivo = 'Datos_CLV_2026_Tec_v2.xlsx'
Metas = pd.read_excel(archivo, sheet_name='Metas')

# Seleccionar solo las variables que utilizaremos
Metas = Metas[['fecha', 'estado', 'meta']]
Metas.head()
```

```python
# Confirmar si hay valores vacíos
print("Valores vacíos:")
print(Metas.isnull().sum())

# Convertir la columna fecha al formato correcto de fecha
Metas['fecha'] = pd.to_datetime(Metas['fecha'], errors='coerce')
```

```python
# Contar la cantidad de registros de meta que tiene cada estado
conteo_estados = Metas['estado'].value_counts()

# Mostrar la tabla en pantalla
print("--- Cantidad de registros por Estado ---")
print(conteo_estados)

# Gráfica de barras horizontales
plt.figure(figsize=(8, 6))
conteo_estados.plot(kind='barh', color='orange')
plt.title('Registros de Metas por Estado')
plt.xlabel('Cantidad de Registros')
plt.ylabel('Estado')
plt.show()
```

```python
# Ver estadísticas básicas de las metas (mínimo, máximo, promedio)
print(Metas['meta'].describe())

# Histograma para ver los valores más comunes de las metas
plt.figure(figsize=(8, 4))
plt.hist(Metas['meta'], bins=15, color='purple', edgecolor='black')
plt.title('Distribución de los Montos de Metas')
plt.xlabel('Monto ($)')
plt.ylabel('Frecuencia')
plt.show()

# Boxplot para ver datos atípicos
plt.figure(figsize=(8, 3))
sns.boxplot(x=Metas['meta'], color='violet')
plt.title('Valores atípicos en las Metas')
plt.xlabel('Monto ($)')
plt.show()
```
